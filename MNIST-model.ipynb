```python
import torchvision
from torchvision import transforms
from torch.utils.data import DataLoader

```

```python
train_data = torchvision.datasets.MNIST("/data/", download = True , train= True)
test_data = torchvision.datasets.MNIST("/data/", download = True , train = False)
train_data
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
t_0 = train_data[1]
print("Image is")
t_0[0]
```

```python
trans = transforms.Compose([transforms.ToTensor()])
```

```python
train_data.transform = trans
test_data.transform = trans


t_0 = train_data[0]
t_0[0].shape
```

```python
batch_size = 32
train_loader  = DataLoader(train_data ,batch_size=batch_size , shuffle= True )
test_loader = DataLoader(test_data , batch_size = batch_size , shuffle = True)

```

```python
import torch.nn as nn 
layers = [
    nn.Flatten(),
    nn.Linear(1*28*28 , 512),# input
    nn.ReLU(),
    nn.Linear(512 , 512 ),
    nn.ReLU(),
    nn.Linear(512 , 10),
    nn.Sigmoid()
]
model = nn.Sequential(*layers)
model.compile()
model.to(device)
```

```python
print(next(model.parameters()).device)
```

```python
loss = torch.nn.CrossEntropyLoss()
optim = torch.optim.Adam(model.parameters() , lr = 0.01)

# training 
epochs = 5
for epoch in range(epochs):
    for x , y in train_loader:
        x = x.to(device)
        y = y.to(device)
        # print(x.device , y.device)
        optim.zero_grad
        out = model(x)
        loss_det = loss(out , y)
        loss_det.backward()
        optim.step()
    print(f"Epoch {epoch+1}, Loss: {loss_det.item()}")
```

```python
from torchmetrics import Accuracy

acc = Accuracy(num_classes=10 , task = 'multiclass' ).to(device)
model.eval()
with torch.no_grad():
    for x , y in test_loader:
        x = x.to(device)
        y = y.to(device)

        out = model(x)
        acu  = acc(out , y)

print(acc.compute().item()*100)
```
